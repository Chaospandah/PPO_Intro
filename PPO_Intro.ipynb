```python
import gymnasium as gym
import mujoco
import stable_baselines3

print("Gymnasium:", gym.__version__)
print("MuJoCo:", mujoco.__version__)
print("Stable-Baselines3:", stable_baselines3.__version__)
```

```python
env = gym.make("Reacher-v5")

print("Observation space:", env.observation_space)
print("Action space:", env.action_space)

print("Observation size:", env.observation_space.shape)
print("Action size:", env.action_space.shape)
```

```python
observation, info = env.reset(seed=42)

print('observation:', observation)
print('number of values:', len(observation))
```

```python
from os import truncate
action = env.action_space.sample()

next_observation, reward, terminated, truncated, info = env.step(action)

print('Action:', action)
print('Reward:', reward)
print('Terminated:', terminated)
print('Truncated:', truncated)
print('Next observation:', next_observation)
```

```python
observation, info = env.reset(seed=42)

total_reward = 0
steps = 0
done = False

while not done:
  action = env.action_space.sample()

  observation, reward, terminated, truncated, info = env.step(action)

  total_reward += reward
  steps += 1

  done = terminated or truncated

print('Total reward:', total_reward)
print('Steps:', steps)
```

```python
import numpy as np

episode_rewards = []

for episode in range(10):
  observation, info = env.reset(seed = episode)
  env.action_space.seed(episode)

  total_reward = 0
  done = False

  while not done:
    action = env.action_space.sample()

    observation, reward, terminated, truncated, info = env.step(action)

    total_reward += reward

    done = terminated or truncated

  episode_rewards.append(total_reward)

print('Episode rewards:', episode_rewards)
print('Mean reward:', np.mean(episode_rewards))
```

```python
from stable_baselines3 import PPO

train_env = gym.make('Reacher-v5')

model = PPO(
    policy='MlpPolicy',
    env = train_env,
    verbose=1,
    device='cpu'
)

print('PPO agent created')
```

```python
model.learn(total_timesteps=100_000)
print('training finished')
```

```python
demo_env = gym.make("Reacher-v5")

observation, info = demo_env.reset(seed=42)

action, _ = model.predict(
    observation,
    deterministic=True
)

next_observation, reward, terminated, truncated, info = demo_env.step(action)

print("PPO action:", action)
print("Reward:", reward)
print("Next observation:", next_observation)
```

```python
eval_env = gym.make('Reacher-v5')

ppo_rewards = []

for episode in range(10):
    observation, info = eval_env.reset(seed=episode)

    total_reward = 0
    done = False

    while not done:
        action, _ = model.predict(
            observation,
            deterministic=True
        )

        observation, reward, terminated, truncated, info = eval_env.step(action)

        total_reward += reward
        done = terminated or truncated

    # This must be outside the while loop
    ppo_rewards.append(total_reward)

print("Number of episodes:", len(ppo_rewards))
print("PPO rewards:", ppo_rewards)
print("PPO mean reward:", np.mean(ppo_rewards))
```

```python
random_mean = -43.808732073026356
ppo_mean = np.mean(ppo_rewards)

improvement = ppo_mean - random_mean

penalty_reduction = (
    (abs(random_mean) - abs(ppo_mean))
    / abs(random_mean)
    * 100
)

print("Random mean:", random_mean)
print("PPO mean:", ppo_mean)
print("Reward improvement:", improvement)
print("Penalty reduction:", penalty_reduction, "%")
```
